```python
from transformers import GPT2LMHeadModel, pipeline, set_seed

model_hf = GPT2LMHeadModel.from_pretrained("gpt2")
statedict = model_hf.state_dict()

for i, v in statedict.items():
    print(i, v.shape)
```

```python
generator = pipeline('text-generation', model='gpt2')
set_seed(42)
output = generator("Hello! I am the best", max_length=10, num_return_sequences=2, do_sample=True)
for x in output:
    print(x['generated_text'])
```
