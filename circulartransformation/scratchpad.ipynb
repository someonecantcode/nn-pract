# Circle generator
We are generating a set of points/vectors that are inside a circle inclusive of its boundaries.


```python
def generate_circle(size: int, r_max: int = 1) -> torch.Tensor:
    values = torch.zeros(size, 2)
    for i in range(size):
        r = random.uniform(0, r_max) 
        theta = random.uniform(0, math.tau) 

        values[i, 0] = r * math.cos(theta);
        values[i, 1] = r * math.sin(theta);

    return values


my_values = generate_circle(10**4);
# for i in range(my_values.size(0)):
#     plt.plot(my_values[i,0], my_values[i,1])
#     plt.show()


plt.scatter(my_values[:,0], my_values[:,1])


```

```python
a = torch.randn(2,2)
b = my_values @ a
# this is dot product

plt.scatter(b[:,0], b[:,1])
```
