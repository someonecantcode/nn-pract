```python
import torch
import matplotlib.pyplot as plt
import numpy as np

print(f"PyTorch version: {torch.__version__}")
print(f"CUDA available: {torch.cuda.is_available()}")
print(f"CUDA device count: {torch.cuda.device_count()}")
if torch.cuda.is_available():
    print(f"Current device: {torch.cuda.get_device_name(0)}")
```

```python
numberofvectors = 100
tendimvector = torch.randn(numberofvectors,10)

U, S, V = torch.pca_lowrank(tendimvector, q = 3, niter=3)

pca_vector = torch.matmul(tendimvector, V)
print(f'original vector shape {tendimvector.shape}')
print(f'pca reduced vector shape {pca_vector.shape}')

fig = plt.figure(figsize=(10,10))
ax = fig.add_subplot(1, 1, 1, projection='3d')

x = pca_vector[:, 0].numpy()
y = pca_vector[:, 1].numpy()
z = pca_vector[:, 2].numpy()
ax.scatter(x, y, z, c=x, cmap='viridis', s=50)

ax.set_title('PCA 3D Projection of 10D Data')
plt.show()
```

```python
#-------------------------------------------------#
# 3d to 2d projection
numberofvectors = 100
vector = torch.randn(numberofvectors, 3)

U, S, V = torch.pca_lowrank(vector, q = 2, niter=3)
pca_vector = torch.matmul(vector, V)

plt.figure()
x = pca_vector[:, 0]
y = pca_vector[:, 1]
plt.scatter(x, y, alpha=0.7, c = x, edgecolors='k')

plt.title('PCA 2D Projection')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```
